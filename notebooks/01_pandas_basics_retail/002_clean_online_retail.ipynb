```python
import pandas as pd
from pathlib import Path
```

```python
project_root = Path.cwd().parent.parent
input_file = project_root / "data" / "raw" / "retail" / "online_retail.csv"

retail_df = pd.read_csv(input_file)
retail_df.head()
```

```python
retail_df.shape
```

```python
retail_df.info()
```

```python
retail_df.isna().sum()
```

```python
clean_df = retail_df.copy()
```

```python
clean_df = clean_df.dropna(subset=["CustomerID", "Description"])
clean_df.shape
```

```python
clean_df["InvoiceDate"] = pd.to_datetime(clean_df["InvoiceDate"])
clean_df["CustomerID"] = clean_df["CustomerID"].astype("int64")
```

```python
clean_df = clean_df[clean_df["Quantity"] > 0]
clean_df = clean_df[clean_df["UnitPrice"] > 0]
clean_df.shape
```

```python
clean_df["Revenue"] = clean_df["Quantity"] * clean_df["UnitPrice"]
clean_df.head()
```

```python
clean_df[["Quantity", "UnitPrice", "Revenue"]].describe()
```

```python
output_dir = project_root / "data" / "cleaned" / "retail"
output_dir.mkdir(parents=True, exist_ok=True)

output_file = output_dir / "online_retail_clean.csv"
clean_df.to_csv(output_file, index=False)

print(f"Clean file saved: {output_file}")
```
