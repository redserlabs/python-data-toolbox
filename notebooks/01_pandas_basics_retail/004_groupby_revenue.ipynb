```python
import pandas as pd
from pathlib import Path
```

```python
project_root = Path.cwd().parent.parent
input_file = project_root / "data" / "cleaned" / "retail" / "online_retail_clean.csv"

df = pd.read_csv(input_file)
df.head()
```

```python
df.shape
```

```python
country_revenue_df = (
    df.groupby("Country", as_index=False)["Revenue"]
    .sum()
    .sort_values(by="Revenue", ascending=False)
)

country_revenue_df.head(10)
```

```python
customer_revenue_df = (
    df.groupby("CustomerID", as_index=False)["Revenue"]
    .sum()
    .sort_values(by="Revenue", ascending=False)
)

customer_revenue_df.head(10)
```

```python
product_revenue_df = (
    df.groupby(["StockCode", "Description"], as_index=False)["Revenue"]
    .sum()
    .sort_values(by="Revenue", ascending=False)
)

product_revenue_df.head(10)
```

```python
country_quantity_df = (
    df.groupby("Country", as_index=False)["Quantity"]
    .sum()
    .sort_values(by="Quantity", ascending=False)
)

country_quantity_df.head(10)
```

```python
invoice_revenue_df = (
    df.groupby("InvoiceNo", as_index=False)["Revenue"]
    .sum()
    .sort_values(by="Revenue", ascending=False)
)

invoice_revenue_df.head(10)
```

```python
country_summary_df = (
    df.groupby("Country", as_index=False)
    .agg(
        total_revenue=("Revenue", "sum"),
        total_quantity=("Quantity", "sum"),
        average_unit_price=("UnitPrice", "mean")
    )
    .sort_values(by="total_revenue", ascending=False)
)

country_summary_df.head(10)
```
