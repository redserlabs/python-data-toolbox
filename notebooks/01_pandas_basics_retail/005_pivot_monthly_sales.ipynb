```python
import pandas as pd
from pathlib import Path
```

```python
project_root = Path.cwd().parent.parent
input_file = project_root / "data" / "cleaned" / "retail" / "online_retail_clean.csv"

df = pd.read_csv(input_file)
df.head()
```

```python
df["InvoiceDate"] = pd.to_datetime(df["InvoiceDate"])
df["InvoiceMonth"] = df["InvoiceDate"].dt.to_period("M").astype(str)

df[["InvoiceDate", "InvoiceMonth"]].head()
```

```python
monthly_revenue_df = (
    df.groupby("InvoiceMonth", as_index=False)["Revenue"]
    .sum()
    .sort_values(by="InvoiceMonth")
)

monthly_revenue_df.head(12)
```

```python
monthly_country_revenue_pivot = pd.pivot_table(
    df,
    values="Revenue",
    index="InvoiceMonth",
    columns="Country",
    aggfunc="sum",
    fill_value=0
)

monthly_country_revenue_pivot.head()
```

```python
monthly_country_revenue_pivot.shape
```

```python
top_countries = (
    df.groupby("Country", as_index=False)["Revenue"]
    .sum()
    .sort_values(by="Revenue", ascending=False)
    .head(5)["Country"]
    .tolist()
)

top_countries
```

```python
top_country_pivot = monthly_country_revenue_pivot[top_countries]
top_country_pivot.head()
```
