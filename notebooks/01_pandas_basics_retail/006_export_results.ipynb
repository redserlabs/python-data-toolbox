```python
import pandas as pd
from pathlib import Path
```

```python
project_root = Path.cwd().parent.parent
input_file = project_root / "data" / "cleaned" / "retail" / "online_retail_clean.csv"

df = pd.read_csv(input_file)
df.head()
```

```python
df["InvoiceDate"] = pd.to_datetime(df["InvoiceDate"])
df["InvoiceMonth"] = df["InvoiceDate"].dt.to_period("M").astype(str)
```

```python
monthly_revenue_df = (
    df.groupby("InvoiceMonth", as_index=False)["Revenue"]
    .sum()
    .sort_values(by="InvoiceMonth")
)

monthly_revenue_df.head()
```

```python
country_revenue_df = (
    df.groupby("Country", as_index=False)["Revenue"]
    .sum()
    .sort_values(by="Revenue", ascending=False)
)

country_revenue_df.head(10)
```

```python
export_dir = project_root / "data" / "exports" / "retail"
export_dir.mkdir(parents=True, exist_ok=True)

monthly_revenue_file = export_dir / "monthly_revenue.csv"
country_revenue_file = export_dir / "country_revenue.csv"
```

```python
monthly_revenue_df.to_csv(monthly_revenue_file, index=False)
country_revenue_df.to_csv(country_revenue_file, index=False)

print(f"File saved: {monthly_revenue_file}")
print(f"File saved: {country_revenue_file}")
```
