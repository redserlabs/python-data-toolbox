```python
import json
import pandas as pd
from pathlib import Path
```

```python
project_root = Path.cwd().parent.parent
input_file = project_root / "data" / "raw" / "api" / "github" / "repo_response.json"

input_file
```

```python
with open(input_file, "r") as file:
    repo_data = json.load(file)

type(repo_data), len(repo_data)
```

```python
repo_data[0]
```

```python
repo_df = pd.DataFrame(repo_data)
repo_df.head()
```

```python
repo_df.shape
```

```python
repo_df.columns.tolist()
```

```python
selected_repo_df = repo_df[
    [
        "id",
        "name",
        "full_name",
        "private",
        "html_url",
        "description",
        "created_at",
        "updated_at",
        "pushed_at",
        "language",
        "visibility"
    ]
]

selected_repo_df.head()
```

```python
selected_repo_df.info()
```

```python
selected_repo_df["created_at"] = pd.to_datetime(selected_repo_df["created_at"])
selected_repo_df["updated_at"] = pd.to_datetime(selected_repo_df["updated_at"])
selected_repo_df["pushed_at"] = pd.to_datetime(selected_repo_df["pushed_at"])

selected_repo_df.dtypes
```
