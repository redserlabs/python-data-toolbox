```python
import json
import pandas as pd
from pathlib import Path
```

```python
project_root = Path.cwd().parent.parent
input_file = project_root / "data" / "raw" / "api" / "github" / "issues_response.json"

input_file
```

```python
with open(input_file, "r") as file:
    issues_data = json.load(file)

type(issues_data), len(issues_data)
```

```python
issues_data[0]
```

```python
normalized_issues = []

for issue in issues_data:
    issue_record = {
        "issue_id": issue["id"],
        "issue_number": issue["number"],
        "title": issue["title"],
        "state": issue["state"],
        "created_at": issue["created_at"],
        "updated_at": issue["updated_at"],
        "user_login": issue["user"]["login"],
        "comments": issue["comments"]
    }
    normalized_issues.append(issue_record)

normalized_issues[:2]
```

```python
issues_df = pd.DataFrame(normalized_issues)
issues_df.head()
```

```python
issues_df.shape
```

```python
issues_df.info()
```

```python
issues_df["created_at"] = pd.to_datetime(issues_df["created_at"])
issues_df["updated_at"] = pd.to_datetime(issues_df["updated_at"])

issues_df.dtypes
```

```python
issues_df.sort_values(by="comments", ascending=False).head(10)
```

```python
output_file = project_root / "data" / "cleaned" / "github_issue_records.csv"
issues_df.to_csv(output_file, index=False)

print(f"File saved: {output_file}")
```
