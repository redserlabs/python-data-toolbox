```python
import pandas as pd
from pathlib import Path
```

```python
project_root = Path.cwd().parent.parent
input_file = project_root / "data" / "cleaned" / "github_issue_records.csv"

issues_df = pd.read_csv(input_file)
issues_df.head()
```

```python
issues_df.info()
```

```python
issues_df["created_at"] = pd.to_datetime(issues_df["created_at"])
issues_df["updated_at"] = pd.to_datetime(issues_df["updated_at"])

issues_df.dtypes
```

```python
issues_df["created_date"] = issues_df["created_at"].dt.date
issues_df["created_hour"] = issues_df["created_at"].dt.hour
issues_df["updated_date"] = issues_df["updated_at"].dt.date
issues_df["update_delay_hours"] = (
    (issues_df["updated_at"] - issues_df["created_at"]).dt.total_seconds() / 3600
)

issues_df.head()
```

```python
issues_by_day_df = (
    issues_df.groupby("created_date", as_index=False)["issue_id"]
    .count()
    .rename(columns={"issue_id": "issue_count"})
    .sort_values(by="created_date")
)

issues_by_day_df
```

```python
issues_by_hour_df = (
    issues_df.groupby("created_hour", as_index=False)["issue_id"]
    .count()
    .rename(columns={"issue_id": "issue_count"})
    .sort_values(by="created_hour")
)

issues_by_hour_df
```

```python
top_users_df = (
    issues_df.groupby("user_login", as_index=False)["issue_id"]
    .count()
    .rename(columns={"issue_id": "issue_count"})
    .sort_values(by="issue_count", ascending=False)
)

top_users_df.head(10)
```

```python
most_commented_issues_df = issues_df.sort_values(by="comments", ascending=False)
most_commented_issues_df.head(10)
```
